```python
#METODOS Y FUNCIONES
# FUNCIÓN
def saludar(nombre):
    return f"Hola, {nombre}" #NO es necesario crear un objeto o clase para usarlo


# CLASE
class Persona:
    #_init_ es para cuando quieres que automaticamente se aplique este metodo al objeto
    def __init__(self, nombre): #SELF es necesario cuando quieres instanciar
        self.nombre = nombre# self es como usar This en c# pero aqui es obigatorio

    # MÉTODO
    def saludar(self):
        return f"Hola, soy {self.nombre}"


# Llamada a la función
print(saludar("Carlos"))

# Creación de un objeto
persona = Persona("Carlos")

# Llamada al método
print(persona.saludar())
```

```python
#class method
class Cuenta:

    cantidad = 0

    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo
        Cuenta.cantidad += 1

    @classmethod #Este es necesario para crear un metodo sin self, no ocupas mencionar al objeto para usarlo
    def consultar_cantidad(cls):
        return cls.cantidad
cuenta1 = Cuenta("Carlos", 5000)
cuenta2 = Cuenta("María", 3000)
cuenta3 = Cuenta("José", 8000)

print(Cuenta.consultar_cantidad())# no usa cuenta1, ni la 2 ni la 3, solo necesita la clase
```

```python
my_list = [8, 10, 6, 2, 4]  # lista a ordenar
swapped = True  # Lo necesitamos verdadero (True) para ingresar al bucle while.

while swapped:
    swapped = False  # no hay intercambios hasta ahora
    for i in range(len(my_list) - 1):
        if my_list[i] > my_list[i + 1]:
            swapped = True  # ¡ocurrió el intercambio!
            my_list[i], my_list[i + 1] = my_list[i + 1], my_list[i]

print(my_list)
```

OJOOO la forma de copiar un arreglo es distinto no sirve hacer new_list=my_list esto solo le daria otro acceso directo al mismo espacio de memoria que my_list, osea que al querer ver new_list veriamos directamente a lo que tiene my_list aun aunque lo modifiques, por eso la forma de copiar arreglos es como la siguiente:

```python
#Esto es lo que pasa si solo se usa el igual
my_list = [10, 8, 6, 4, 2]
new_list = my_list
new_list[0]=1
print(my_list)
print(new_list)
```

```python
#Esto es lo que pasa si se usa el [:] para copiar
my_list = [10, 8, 6, 4, 2]
new_list = my_list[1:3]#copia solo el valor al que pertene el 1 hasta un numero antes de 3 osea solo copia el valor 1 y 2 en el indice del arreglo
new_list2=my_list[:]#copia el arreglo completamente en otro arreglo
print(new_list)
print(new_list2)

```

```python
#OJOOOOO IMPORTANTEEEE SE PUEDEN COMPARAR VALORES DIRECTAMENTE CON EL FOR
drawn = [5, 11, 9, 42, 3, 49]
bets = [3, 7, 11, 42, 34, 49]
hits = 0

for number in bets: #Number toma el valor de la posicion correspondiente de la lista bets
    if number in drawn: #se busca el valor de number en drawn
        hits += 1

print(hits)
```

OJOOOOO importante COMPRENSION DE LISTA
El siguiente codigo hace que la lista squares se llene de el cuadrado de los numeros del 0 al 9

```python
#OJOOOOO IMPORTANTEEEEEEE
squares = [x ** 2 for x in range(10)]
print(squares)
```

```python
twos = [2 ** i for i in range(8)]
#AQUI seran las potencias del numero 2 porque el valor que cambia esta ves es la i o la potencia
print(twos)
```

```python
odds = [x for x in squares if x % 2 != 0 ]
print(odds)
#crea una lista con solo los elementos impares de la lista squares.
```
